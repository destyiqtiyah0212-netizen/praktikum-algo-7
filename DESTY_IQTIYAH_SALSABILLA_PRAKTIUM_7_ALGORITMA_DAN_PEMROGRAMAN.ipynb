```python
def ordinal(angka):
    """Return tuple (angka, akhiran ordinal)."""
    if angka % 100 in (11, 12, 13):
        akhiran = "th"
    elif angka % 10 == 1:
        akhiran = "st"
    elif angka % 10 == 2:
        akhiran = "nd"
    elif angka % 10 == 3:
        akhiran = "rd"
    else:
        akhiran = "th"
    return angka, akhiran


print("Ordinal Number")
print("ketik 0 untuk mengentikan program")
while True:
    try:
        angka = int(input("masukkan angka: "))
    except ValueError:
        print("Input harus berupa angka")
        continue

    print(ordinal(angka))
    if angka == 0:
        print("terima kasih telah menggunakan program saya")
        break
```

```python
def cek_prima(n):
    """Return True jika n bilangan prima."""
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True


def tampilkan_hasil(n):
    """Cetak hasil pengecekan bilangan prima."""
    if cek_prima(n):
        print(n, "adalah bilangan Prima")
    else:
        print(n, "bukanlah bilangan Prima")


angka = int(input("Masukkan angka: "))
tampilkan_hasil(angka)
```
